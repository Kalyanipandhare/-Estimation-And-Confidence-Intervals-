```python
import numpy as np
import scipy.stats as stats

# Sample data (durability in millions of characters)
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

# Sample size
n = len(data)

# Sample mean and standard deviation
mean = np.mean(data)
std_dev = np.std(data, ddof=1)  # ddof=1 for sample standard deviation

# Confidence level
confidence = 0.99
alpha = 1 - confidence
df = n - 1  # degrees of freedom

# t-critical value
t_critical = stats.t.ppf(1 - alpha/2, df)

# Margin of Error
margin_error = t_critical * (std_dev / np.sqrt(n))

# Confidence interval
lower_bound = mean - margin_error
upper_bound = mean + margin_error

# Display results
print(f"Sample Mean: {mean:.3f}")
print(f"Sample Standard Deviation: {std_dev:.3f}")
print(f"t-critical value (99% confidence): {t_critical:.3f}")
print(f"Margin of Error: {margin_error:.3f}")
print(f"99% Confidence Interval: ({lower_bound:.3f}, {upper_bound:.3f})")
```

```python
"""
Estimation of the mean durability of print-heads using a small sample size (n = 15).

 t-distribution is used because the sample is small and population standard deviation is unknown.

Answer

Mean durability = 1.239 million characters

99% Confidence Interval: (1.090, 1.387)

Interpretation:
We are 99% confident that the true population mean durability lies within this interval."""

```

```python
import numpy as np
from scipy.stats import norm

# Given values
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
n = len(data)
mean = np.mean(data)
sigma = 0.2  # known population standard deviation

# Z-critical value for 99% confidence
z_critical = norm.ppf(0.995)  # 0.5% in each tail

# Margin of Error
margin_error = z_critical * (sigma / np.sqrt(n))

# Confidence interval
lower_bound = mean - margin_error
upper_bound = mean + margin_error

# Display results
print(f"Sample Mean: {mean:.3f}")
print(f"Known Population Std Dev: {sigma}")
print(f"Z-critical value (99% confidence): {z_critical:.3f}")
print(f"Margin of Error: {margin_error:.3f}")
print(f"99% Confidence Interval: ({lower_bound:.3f}, {upper_bound:.3f})")
```
